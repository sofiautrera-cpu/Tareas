Filtrado de Datos:
- Filtra el DataFrame para mostrar solo las filas donde las ventas en América del Norte (NA_Sales) sean mayores a 1 millón.



```python
# Filtrar juegos con más de 1 millón en ventas en Norteamérica
# En el dataset, las ventas están en millones de unidades, así que “mayores a 1 millón” significa:
# NA_Sales > 1.0
df_na_gt1 = df[df["NA_Sales"] > 1.0]
print(df_na_gt1.head(10))
print(f"Total de juegos con más de 1M en NA: {len(df_na_gt1)}")
```

¿Qué hace este código?
- df["NA_Sales"] > 1.0 → crea un filtro booleano
- df[...] → devuelve solo las filas que cumplen la condición
- df_na_gt1 → DataFrame filtrado
- len(df_na_gt1) → cuántos juegos superan 1M en NA

- Filtra el DataFrame para mostrar solo las filas donde las ventas en Japón (JP_Sales) sean menores a 0.1 millón.


```python
# Filtrar juegos con menos de 0.1M en ventas en Japón
# La condición es: JP_Sales < 0.1

df_jp_lt_01 = df[df["JP_Sales"] < 0.1]

print(df_jp_lt_01.head(10))
print(f"Total de juegos con menos de 0.1M en Japón: {len(df_jp_lt_01)}")
```

- ¿Qué hace este código?
- df["JP_Sales"] < 0.1 → crea un filtro booleano
- df[...] → devuelve solo las filas que cumplen la condición
- df_jp_lt_01 → DataFrame filtrado
- len(df_jp_lt_01) → cuántos juegos venden menos de 100k en Japón

- Utilizando el método query(), filtra el DataFrame para mostrar las filas donde el género sea Action y las ventas globales (Global_Sales) sean mayores a 2 millones.

```python
# Filtrar juegos de género Action con ventas globales > 2 millones
# La condición es:
# Genre == "Action"
# Global_Sales > 2

df_action_gt2 = df.query('Genre == "Action" and Global_Sales > 2')

print(df_action_gt2.head(10))
print(f"Total de juegos Action con más de 2M en ventas globales: {len(df_action_gt2)}")
```

- ¿Qué hace este código?
- query() permite escribir condiciones como si fueran expresiones lógicas normales.
- Genre == "Action" filtra solo juegos de acción.
- Global_Sales > 2 selecciona los que superan los 2 millones de ventas globales.
- df_action_gt2 contiene el resultado filtrado.

Slicing de Datos:
- Selecciona y muestra solo las columnas Name y Global_Sales del DataFrame.


```python
# Seleccionar solo las columnas Name y Global_Sales
# Puede hacerse de dos formas: con lista de columnas o con notación de doble corchete.

# Opción 1 (recomendada, más explícita)

df_slice = df[["Name", "Global_Sales"]]
print(df_slice.head(10))
```

```python
# Opción 2 (equivalente)

print(df[["Name", "Global_Sales"]].head(10))
```

- ¿Qué hace este slicing?
- df[["Name", "Global_Sales"]] selecciona únicamente esas dos columnas.
- head(10) muestra las primeras 10 filas para inspección rápida.

- Utilizando loc[], selecciona y muestra las filas de la 5 a la 10 (inclusive) y las columnas Name y Genre.


```python
# Vamos a hacer slicing con loc[], seleccionando:
# Filas 5 a 10 (ambas incluidas)
# Columnas: Name y Genre

# Código con loc[]
df_loc_slice = df.loc[5:10, ["Name", "Genre"]]
print(df_loc_slice)
```

- ¿Qué hace este slicing?
- 5:10 → incluye las filas desde la 5 hasta la 10.
- ["Name", "Genre"] → selecciona solo esas dos columnas.
- loc[] → slicing basado en etiquetas, por eso incluye el índice final.

- Utilizando iloc[], selecciona y muestra las primeras 5 filas y las primeras 3 columnas del DataFrame.

```python
# Con iloc[], que selecciona por posición numérica (no por etiquetas).
# Condiciones:
# Primeras 5 filas → posiciones 0:5
# Primeras 3 columnas → posiciones 0:3

# Código con iloc[]
df_iloc_slice = df.iloc[0:5, 0:3]
print(df_iloc_slice)
```

- ¿Qué hace este slicing?
- 0:5 → filas desde la posición 0 hasta la 4 (5 filas en total).
- 0:3 → columnas 0, 1 y 2.
- iloc[] → slicing basado en índices numéricos, por eso el límite superior no se incluye.
