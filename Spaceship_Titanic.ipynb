```python
# 2. Limpiar valores faltantes (Preprocesamiento)
# Objetivo: Imputar valores faltantes según el tipo de variable
# Manejar columnas con muchos NaN (Cabin) mediante extracción de componentes
# Preparar el dataset para el pipeline de modelado

# 2.1 Identificar valores faltantes
df.isna().sum()
# 2.2 Función de preprocesamiento (modular y reutilizable)
def preprocess(df):
    # ============================
    # 1. Manejo de Cabin (muchos NaN)
    # ============================
    # Cabin tiene formato: Deck/Num/Side
    df[['Deck', 'Num', 'Side']] = df['Cabin'].str.split('/', expand=True)

    # Imputar componentes de Cabin
    df['Deck'].fillna(df['Deck'].mode()[0], inplace=True)
    df['Side'].fillna(df['Side'].mode()[0], inplace=True)

    # ============================
    # 2. Crear variable TotalSpent
    # ============================
    gastos = ['RoomService', 'FoodCourt', 'ShoppingMall', 'Spa', 'VRDeck']
    df['TotalSpent'] = df[gastos].sum(axis=1)

    # Imputar gastos faltantes con 0
    df[gastos] = df[gastos].fillna(0)

    # ============================
    # 3. Imputación de variables categóricas
    # ============================
    cat_cols = ['HomePlanet', 'CryoSleep', 'Destination', 'VIP']

    for col in cat_cols:
        df[col].fillna(df[col].mode()[0], inplace=True)

    # ============================
    # 4. Imputación de variables numéricas
    # ============================
    num_cols = ['Age', 'TotalSpent']

    for col in num_cols:
        df[col].fillna(df[col].median(), inplace=True)

    # ============================
    # 5. Conversión de tipos
    # ============================
    df['CryoSleep'] = df['CryoSleep'].astype(bool)
    df['VIP'] = df['VIP'].astype(bool)

    return df
# 2.3 Aplicar preprocesamiento al dataset
df_clean = preprocess(df)
df_clean.head()
# 2.4 Verificar que no quedan valores faltantes
df_clean.isna().sum()
```

- Categóricos → modo

```python
# 2. Limpieza y Feature Engineering (Código Completo)

# ============================
# 2. Limpieza y Feature Engineering
# ============================

def clean_and_engineer(df):
    # ----------------------------------------
    # 1. Separar Cabin → Deck, Num, Side
    # ----------------------------------------
    # Cabin tiene formato: Deck/Num/Side
    df[['Deck', 'Num', 'Side']] = df['Cabin'].str.split('/', expand=True)

    # Imputar Deck y Side con el modo
    df['Deck'].fillna(df['Deck'].mode()[0], inplace=True)
    df['Side'].fillna(df['Side'].mode()[0], inplace=True)

    # ----------------------------------------
    # 2. Crear TotalSpent
    # ----------------------------------------
    gastos = ['RoomService', 'FoodCourt', 'ShoppingMall', 'Spa', 'VRDeck']

    # Reemplazar NaN por 0 en gastos
    df[gastos] = df[gastos].fillna(0)

    # Crear variable TotalSpent
    df['TotalSpent'] = df[gastos].sum(axis=1)

    # ----------------------------------------
    # 3. Imputación de valores faltantes
    # ----------------------------------------

    # Categóricas → modo
    cat_cols = ['HomePlanet', 'CryoSleep', 'Destination', 'VIP']
    for col in cat_cols:
        df[col].fillna(df[col].mode()[0], inplace=True)

    # Numéricas → mediana
    num_cols = ['Age', 'TotalSpent']
    for col in num_cols:
        df[col].fillna(df[col].median(), inplace=True)

    # ----------------------------------------
    # 4. Conversión de tipos
    # ----------------------------------------
    df['CryoSleep'] = df['CryoSleep'].astype(bool)
    df['VIP'] = df['VIP'].astype(bool)

    return df


# Aplicar limpieza al dataset
df_clean = clean_and_engineer(df)

# Verificar resultado
df_clean.head()
df_clean.isna().sum()
```
