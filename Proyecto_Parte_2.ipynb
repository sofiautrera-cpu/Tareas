```python
 # Instrucciones:

# Transformación de Datos
# Crea nuevas columnas: Basándonos en los datos existentes, crea nuevas columnas que sean útiles para el análisis. Por ejemplo, calcula el ingreso total por venta y normaliza las ventas.

# A) Calcula el ingreso total por venta

# B) Normaliza las ventas usando Min–Max
# Incluye validación para evitar errores si faltan columnas

# Función modular y validada

def crear_nuevas_columnas(df):
    """
    Crea nuevas columnas útiles para el análisis:
    - total_income: ingreso total por venta (price per unit* Quantity)
    - sales_norm: normalización Min-Max de la columna 'Total Amount'
    """

    # --- 1. Ingreso total por venta ---
    if {"Price per Unit", "Quantity"}.issubset(df.columns):
        df["total_income"] = df["Price per Unit"] * df["Quantity"]
    else:
        print("⚠️ No se creó 'total_income': faltan columnas 'Price per Unit' o 'Quantity'.")

    # --- 2. Normalización de ventas (Total Amount) ---
    if "Total Amount" in df.columns:
        min_val = df["Total Amount"].min()
        max_val = df["Total Amount"].max()

        if max_val > min_val:  # evita división por cero
            df["sales_norm"] = (df["Total Amount"] - min_val) / (max_val - min_val)
        else:
            print("⚠️ No se normalizó 'Total Amount': todos los valores son iguales.")
    else:
        print("⚠️ No se creó 'sales_norm': falta la columna 'Total Amount'.")

    return df # Explicitly return the modified DataFrame

df = crear_nuevas_columnas(df)
df.head()
```

```python
# Clasifica los datos: Crea una columna que clasifique las ventas en categorías significativas (e.g., ‘Alta’, ‘Media’, ‘Baja’).

# Definir una función para clasificar las ventas
def classify_sales(amount):
    if amount <= df['Total Amount'].quantile(0.33):
        return 'Baja'
    elif amount <= df['Total Amount'].quantile(0.66):
        return 'Media'
    else:
        return 'Alta'

# Aplicar la función para crear la nueva columna 'Sales_Category'
df['Sales_Category'] = df['Total Amount'].apply(classify_sales)

# Mostrar el conteo de cada categoría para verificar
print("Conteo de 'Sales_Category':")
print(df['Sales_Category'].value_counts())

# Mostrar las primeras filas del DataFrame con la nueva columna
print("\nDataFrame con 'Sales_Category':")
print(df.head())
```

```python
# Agrupación y Agregación:
# Agrupación por múltiples columnas: Realiza agrupaciones por categorías como producto y tienda, producto y mes, etc.

# Agrupar por 'Product Category' y 'Store' y calcular la suma de 'Total Amount' y la media de 'Quantity'
print("Agrupación por 'Product Category' y 'Store':")
grouped_data = df.groupby(['Product Category', 'Store']).agg(
    total_ventas=('Total Amount', 'sum'),
    cantidad_promedio=('Quantity', 'mean')
)
print(grouped_data)

# Agrupar por 'Product Category' y 'Gender' y calcular el total de 'Total Amount'
print("\nAgrupación por 'Product Category' y 'Gender':")
grouped_gender = df.groupby(['Product Category', 'Gender'])['Total Amount'].sum()
print(grouped_gender)
```

```python
# Aplicar funciones de agregación: Utiliza funciones como sum, mean, count, min, max, std, y var para obtener estadísticas descriptivas de cada grupo.

# Agrupar por 'Product Category' y aplicar múltiples funciones de agregación a 'Total Amount'
print("Estadísticas descriptivas de 'Total Amount' por 'Product Category':")
agregaciones_por_categoria = df.groupby('Product Category')['Total Amount'].agg([
    'sum', 'mean', 'count', 'min', 'max', 'std', 'var'
])
print(agregaciones_por_categoria)

# También podemos aplicar diferentes agregaciones a diferentes columnas
print("\nEstadísticas variadas por 'Product Category':")
agregaciones_variadas = df.groupby('Product Category').agg(
    Total_Sales=('Total Amount', 'sum'),
    Average_Price=('Price per Unit', 'mean'),
    Number_of_Transactions=('Transaction ID', 'count')
)
print(agregaciones_variadas)
```

```python
# Análisis Personalizado con apply
# Función personalizada: Aplica funciones personalizadas para realizar análisis específicos que no se pueden lograr con las funciones de agregación estándar.

# Análisis Personalizado con apply()
# Las funciones de agregación estándar (sum, mean, count, etc.) no siempre permiten capturar patrones complejos. Para esos casos, usamos funciones personalizadas aplicadas con apply(), lo que nos da flexibilidad total para crear métricas avanzadas.

# Función modular + validación + documentación

def analisis_personalizado(df):
    """
    Aplica funciones personalizadas con apply() para generar análisis avanzados
    que no pueden lograrse con agregaciones estándar.

    Operaciones realizadas:
    - Crea una columna 'score_calidad' basada en una función personalizada que
      evalúa múltiples variables.
    - Crea una columna 'ratio_precio' que compara el precio con el promedio global.
    - Aplica una función fila a fila para detectar productos destacados.

    Parámetros
    ----------
    df : pandas.DataFrame
        DataFrame con columnas relevantes como 'Price per Unit', 'rating', 'Ventas'.

    Retorna
    -------
    pandas.DataFrame
        DataFrame con nuevas columnas derivadas del análisis personalizado.
    """

    # --- 1. Función personalizada para evaluar calidad ---
    def calcular_score(row):
        """
        Calcula un puntaje compuesto basado en rating y ventas.
        """
        rating = row.get("rating", None)
        sales = row.get("Ventas", None)

        if rating is None or sales is None:
            return None

        # Fórmula personalizada: ponderación entre rating y ventas
        return (rating * 0.7) + (sales * 0.3)

    df["score_calidad"] = df.apply(calcular_score, axis=1)

    # --- 2. Ratio de precio respecto al promedio ---
    if "Price per Unit" in df.columns:
        precio_promedio = df["Price per Unit"].mean()
        df["ratio_precio"] = df["Price per Unit"].apply(lambda p: p / precio_promedio)
    else:
        print("⚠️ No se creó 'ratio_precio': falta la columna 'Price per Unit'.")

    # --- 3. Detección de productos destacados ---
    def es_destacado(row):
        """
        Marca como destacado si tiene alto rating y alto score.
        """
        if row.get("rating") is None or row.get("score_calidad") is None:
            return False
        return row["rating"] > 4.5 and row["score_calidad"] > df["score_calidad"].mean()

    df["destacado"] = df.apply(es_destacado, axis=1)

    return df

df = analisis_personalizado(df)
df.head()
```

- Qué aporta este análisis personalizado
- Combina múltiples variables
Permite crear métricas compuestas como score_calidad.

- Detecta patrones no evidentes
Por ejemplo, productos destacados según criterios propios.

- Normaliza y compara
El ratio_precio permite comparar precios entre productos.

- Flexibilidad total
- apply() permite lógica compleja que no se puede lograr con groupby().agg().

- Análisis personalizado con apply():  
- Para capturar patrones complejos que no pueden obtenerse mediante funciones de agregación estándar, se utilizaron funciones personalizadas aplicadas con apply().
- Estas funciones permiten evaluar múltiples variables simultáneamente, generar métricas compuestas y detectar elementos destacados dentro del dataset.

- Entre las transformaciones realizadas se incluyen:

- Cálculo de un puntaje compuesto de calidad (score_calidad) basado en rating y ventas.

- Creación de un indicador relativo de precio (ratio_precio) respecto al promedio global.

- Identificación de productos destacados mediante una función fila a fila.

- Este enfoque proporciona un análisis más profundo y flexible, adaptado a las necesidades específicas del estudio.

```python
# Ejemplo de uso avanzado: Calcula la desviación de cada venta respecto a la media de su grupo.

# Calcular la media de 'Total Amount' para cada 'Product Category'
mean_total_amount_by_category = df.groupby('Product Category')['Total Amount'].transform('mean')
```
