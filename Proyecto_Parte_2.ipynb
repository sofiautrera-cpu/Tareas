```python
# Asignar aleatoriamente 5 tiendas a los datos del dataset
import numpy as np

# Definir las 5 tiendas
tiendas = ["Store_A", "Store_B", "Store_C", "Store_D", "Store_E"]

# Asignar aleatoriamente una tienda a cada fila del DataFrame
df["Store"] = np.random.choice(tiendas, size=len(df))

# Verificar resultado
print(df["Store"].unique())
print(df.head())
```

# **5. Filtrado de Datos**



```python
# Filtra el DataFrame para mostrar solo las filas donde las ventas (Total Amount) sean mayores a 50.
# Filtrar el DataFrame donde 'Total Amount' sea mayor a 50
df_ventas_gt50 = df[df["Total Amount"] > 50]

print(df_ventas_gt50.head())
print(f"Total de filas con 'Total Amount' > 50: {len(df_ventas_gt50)}")
```

```python
# Filtra el DataFrame para mostrar solo las filas donde el precio (Price per Unit) sea menor a 0.5.
df_precio_lt_05 = df[df["Price per Unit"] < 0.5]
print(df_precio_lt_05.head())
print(f"Total de filas con Price per Unit < 0.5: {len(df_precio_lt_05)}")
```

```python
# Utilizando el método query(), filtra el DataFrame para mostrar las filas donde el producto sea 'Clothing' y las ventas (Total Amount) sean mayores a 30.
df_manzanas_gt30 = df.query('`Product Category` == "Clothing" and `Total Amount` > 30')

print(df_manzanas_gt30.head())
print(f"Total de filas con Product Category 'Clothing' y Total Amount > 30: {len(df_manzanas_gt30)}")
```

## **6. Slicing de Datos**


```python
# Selecciona y muestra solo las columnas Producto y Ventas del DataFrame.
df_seleccion = df[["Product Category", "Total Amount"]]
print(df_seleccion.head())
```

```python
# Utilizando loc[], selecciona y muestra las filas de la 5 a la 10 (inclusive) y las columnas Producto y Tienda.
df_loc_slice = df.loc[5:10, ["Product Category", "Store"]]
print(df_loc_slice)
```

```python
# Utilizando iloc[], selecciona y muestra las primeras 5 filas y las primeras 3 columnas del DataFrame.
df_iloc_slice = df.iloc[0:5, 0:3]
print(df_iloc_slice)
```
